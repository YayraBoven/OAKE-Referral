**Phase 1 : Encoding & Final Cleaning**

After assessing the information
about our three datasets, we start cleaning and encoding them.

In this phase, we do not alter the APR Severity of Illness Description, APR Risk of Mortality and Patient disposition columns of the Hospital Inpatient Discharge Dataset.
This is because the original values are needed in our Phase 2 to create our suitable_for_downward_referral target.



```python
#Encoding rcount, gender and facid because they are categorical and the model needs to work numerically.
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

categorical_columns = ["gender", "facid"]

for col in categorical_columns:
    length_of_stay[col] = encoder.fit_transform(length_of_stay[col])


# We replace "5+" (patients with 5 or more previous counts) with 5
length_of_stay["rcount"] = length_of_stay["rcount"].replace("5+", 5)

# We also convert rcount from object values to integers
length_of_stay["rcount"] = length_of_stay["rcount"].astype(int)
```

```python
#We create a copy, to prevent altering the original dataset.
hosp_discharge_df = hosp_inpatient_discharge.copy()

# Here, duplicate records are removed to produce optimal performance in the model.
hosp_discharge_df = hosp_discharge_df.drop_duplicates()

# We also drop columns that are not useful for the referral model such as administrative columns.
#Currently, our model focuses on informayion relating to the patient's clinical information, treatment etc.
#Unrelated columns could add noise and cause the model to learn hospital-specific patterns.
columns_to_drop = [
    'Hospital Service Area',
    'Hospital County',
    'Operating Certificate Number',
    'Permanent Facility Id',
    'Facility Name',
    'Zip Code - 3 digits',
    'Discharge Year',
    'Payment Typology 1',
    'Payment Typology 2',
    'Payment Typology 3',
    'Birth Weight',
    'Total Charges',
    'Total Costs',
    'Race',
    'Ethnicity'
]

hosp_discharge_df = hosp_discharge_df.drop(columns=columns_to_drop)
```

```python
import os
#The length of stay values are converted from objects to numbers and invalid values are changed to NaN  using errors = ;coerce'
hosp_discharge_df['Length of Stay'] = pd.to_numeric(
    hosp_discharge_df['Length of Stay'],
    errors='coerce'
)
# We store the APR code columns as nullable integers, which allow missing values.
hosp_discharge_df['APR DRG Code'] = hosp_discharge_df['APR DRG Code'].astype('Int64')
hosp_discharge_df['APR MDC Code'] = hosp_discharge_df['APR MDC Code'].astype('Int64')
hosp_discharge_df['APR Severity of Illness Code'] = hosp_discharge_df['APR Severity of Illness Code'].astype('Int64')

#Here we identify the categorical columns and make sure that spaces don't prevent the identification of the same value
categorical_columns = hosp_discharge_df.select_dtypes(
    include='object'
).columns

for col in categorical_columns:
    hosp_discharge_df[col] = hosp_discharge_df[col].str.strip()


#This part removes records with no diagnosis
hosp_discharge_df = hosp_discharge_df.dropna(subset=['CCSR Diagnosis Code', 'CCSR Diagnosis Description'])
#This part removes records with no lenth of stay value
hosp_discharge_df = hosp_discharge_df.dropna(subset=['Length of Stay'])

#Here, we treat missing procedures as no recorded procedure, using "None"
hosp_discharge_df['CCSR Procedure Code'] = (hosp_discharge_df['CCSR Procedure Code'].fillna('None'))
hosp_discharge_df['CCSR Procedure Description'] = (hosp_discharge_df['CCSR Procedure Description'].fillna('None'))


#We remove missing severity and mortality risk values due to their importance
hosp_discharge_df = hosp_discharge_df.dropna(subset=['APR Severity of Illness Description', 'APR Risk of Mortality'])


os.makedirs('data/processed', exist_ok=True)

#We save the cleaned dataset to prevent extra pre-processing
hosp_discharge_df.to_csv('data/processed/hospital_discharge_cleaned.csv',index=False)

#check the result
print(hosp_discharge_df.shape)

print(hosp_discharge_df.isnull().sum())

print(hosp_discharge_df.dtypes)
```

```python
# In this cell, we mainly drop redundant or description columns.
hosp_discharge_df = hosp_discharge_df.drop(columns=[
    'CCSR Diagnosis Description',
    'CCSR Procedure Description',
    'APR DRG Description',
    'APR MDC Description'
])

print(hosp_discharge_df.shape)
```

```python
# We use LabelEncoder to encode remaining categorical columns so they are useful to our model
from sklearn.preprocessing import LabelEncoder

# For Age Groups , we assign values that follow the natiral progression from young to old
age_order = {'0 to 17': 0, '18 to 29': 1, '30 to 49': 2, '50 to 69': 3, '70 or Older': 4}
hosp_discharge_df['Age Group'] = hosp_discharge_df['Age Group'].map(age_order)

# For Gender, we use ne-hot (M/F/U, only 3 categories)
hosp_discharge_df = pd.get_dummies(hosp_discharge_df, columns=['Gender'], prefix='Gender')

# Type of Admission -> one-hot (6 categories, no natural order)
hosp_discharge_df = pd.get_dummies(hosp_discharge_df, columns=['Type of Admission'], prefix='Admission')

# CCSR Diagnosis / Procedure Code -> label encode (too high-cardinality for one-hot: 477 / 321 categories)
le_diag = LabelEncoder()
hosp_discharge_df['CCSR Diagnosis Code'] = le_diag.fit_transform(hosp_discharge_df['CCSR Diagnosis Code'])

le_proc = LabelEncoder()
hosp_discharge_df['CCSR Procedure Code'] = le_proc.fit_transform(hosp_discharge_df['CCSR Procedure Code'])

# APR Medical Surgical Description -> binary
hosp_discharge_df['APR Medical Surgical Description'] = hosp_discharge_df['APR Medical Surgical Description'].map(
    {'Medical': 0, 'Surgical': 1}
)

# Emergency Department Indicator -> binary
hosp_discharge_df['Emergency Department Indicator'] = hosp_discharge_df['Emergency Department Indicator'].map(
    {'N': 0, 'Y': 1}
)

print(hosp_discharge_df.shape)
print(hosp_discharge_df.dtypes)
```

```python
# This function simplifies the hospital-level descriptions into 3 tiers
#For easier recommendations and identification
def map_tier(level):
    level = level.lower()
    if 'teaching' in level or 'quaternary' in level or 'regional' in level:
        return 1  # Major referral-level
    elif 'district' in level or 'municipal' in level or 'secondary' in level:
        return 2  # Mid-level
    else:
        return 3  # Primary/Community

hospitals['hospital_tier'] = hospitals['hospital_level'].apply(map_tier)
print(hospitals[['hospital_name', 'hospital_level', 'hospital_tier']])

hospitals.to_csv('data/processed/hospitals_phase1.csv', index=False)
print("Saved data/processed/hospitals_phase1.csv")
```

**Phase 2: Target Variable Engineering**

Patient Disposition is used to create our target because it is the closest to an outcome.
APR Severity of Illness and APR Risk of Mortality are clinical assessments (available at time of care), so they become input features here, not part of the label. This avoids the model leaking its own answer.

**suitable_for_downward_referral = 1**
Stepped down to a lower-level hospital.

**suitable_for_downward_referral = 0**
Still in need of  continued high-level/specialized care

**Excluded from labeled data** Administrative/legal, not a clinical stability judgment

```python
#These are outcomes that represent a downward referral for less-severe care
positive = ['Home or Self Care', 'Home w/ Home Health Services', 'Skilled Nursing Home',
            'Inpatient Rehabilitation Facility', 'Facility w/ Custodial/Supportive Care',
            'Medicare Cert Long Term Care Hospital', 'Hosp Basd Medicare Approved Swing Bed',
            'Medicaid Cert Nursing Facility', 'Critical Access Hospital']
#These are the outcomes that suggest a patient cannot be downward-referred and still needs specialised care.
negative = ['Expired', 'Short-term Hospital', 'Psychiatric Hospital or Unit of Hosp',
            'Hospice - Home', 'Hospice - Medical Facility', "Cancer Center or Children's Hospital"]

#These do not really show any information concerning patient stability
excluded = ['Left Against Medical Advice', 'Court/Law Enforcement',
            'Another Type Not Listed', 'Federal Health Care Facility']
#We drop these columns here
hosp_discharge_df = hosp_discharge_df[~hosp_discharge_df['Patient Disposition'].isin(excluded)]

#Here, a binary target is created where 1 means suitable for downward referral and 0 means not suitable
hosp_discharge_df['suitable_for_downward_referral'] = hosp_discharge_df['Patient Disposition'].apply(
    lambda x: 1 if x in positive else 0
)
#These 2 lines check the number and percentage in the target classes.
print(hosp_discharge_df['suitable_for_downward_referral'].value_counts())
print(hosp_discharge_df['suitable_for_downward_referral'].value_counts(normalize=True).round(4) * 100)
```

The target is imbalanced with about 92.4% referrable and 7.6% not referrable. This reflects real clinical situations where most admissions end in routine discharge.

Due to this we may need class weighting and evaluate with precision/recall/F1.

```python
# We encode Severity of Illness and Risk of Mortality ordinally, due to the higher progression in them.
severity_order = {'Minor': 0, 'Moderate': 1, 'Major': 2, 'Extreme': 3}
hosp_discharge_df['APR Severity of Illness Description'] = hosp_discharge_df['APR Severity of Illness Description'].map(severity_order)

risk_order = {'Minor': 0, 'Moderate': 1, 'Major': 2, 'Extreme': 3}
hosp_discharge_df['APR Risk of Mortality'] = hosp_discharge_df['APR Risk of Mortality'].map(risk_order)
#We rename the columns to properly show the encoded values
hosp_discharge_df = hosp_discharge_df.rename(columns={'APR Risk of Mortality': 'APR Risk of Mortality Code'})

#The APR Severity of Illness Code is now redundant with the Description we just encoded, so we drop duplicates
hosp_discharge_df = hosp_discharge_df.drop(columns=['APR Severity of Illness Description'])

# Patient Disposition was used to creaet the target, so keeping it as a feature would be a leakage, so it is dropped.
hosp_discharge_df = hosp_discharge_df.drop(columns=['Patient Disposition'])

print(hosp_discharge_df.shape)
print(hosp_discharge_df.columns.tolist())
print("Nulls:", hosp_discharge_df.isnull().sum().sum())
```

```python
hosp_discharge_df.to_csv('data/processed/hosp_discharge_phase2_final.csv', index=False)
print("Saved data/processed/hosp_discharge_phase2_final.csv -- ready for Phase 3 modeling")
```

length_of_stay role for Phase 3

This dataset does not have a column that works like a disposition outcome. There is nothing similar to Patient Disposition that could be used to create a comparable label. If we forced an artificial proxy, like setting a cutoff point on lengthofstay, it would add weak and unreliable labels into the training process.
Decision: hosp_discharge_df, which comes from SPARCS, will be used as the main training set for the classifier, since it contains real outcome data that the model can learn from.length_of_stay will be treated as a secondary and exploratory resource. It is useful for checking how comorbidities and lab values relate to length of stay, but it will not be used to train the target variable for the main classifier.
