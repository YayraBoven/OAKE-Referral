```python
#Inspecting hospitals
# Displays the names of all the columns in the dataset.
print("COLUMN NAMES")
print(hospitals.columns)

# Returns the number of rows and columns in the dataset.
print("DATASET SHAPES")
print(hospitals.shape)


# Displays the first five rows to give an overview of the dataset.
print("FIRST FIVE ROWS")

print(hospitals.head())

# Shows information about the dataset, including data types, non-null values, and memory usage.
print("DATASET INFORMATION")
hospitals.info()

# Returns summary statistics for all numerical columns in the dataset.
print("SUMMARY STATISTICS")
print(hospitals.describe())

# Counts the number of missing values in each column.
print("MISSING VALUES")
print(hospitals.isna().sum())

# Counts the total number of duplicate rows in the dataset.
print("DUPLICATE ROWS")
print(hospitals.duplicated().sum())

# Displays the data type of each column.
print("DATA TYPES")
print(hospitals.dtypes)

#Checking the uniqueness of the hospitals
print("UNIQUENESS OF HOSPITALS")
print(hospitals["hospital_id"].is_unique)

#Number of unique ids
print("NUMBER OF UNIQUE IDs:")
print(hospitals["hospital_id"].nunique())

#Number of hospitals
print("NUMBER OF HOSPITALS:")
print(len(hospitals))

```

```python
#Checking for text inconsistencies

#For regions
print("\nREGIONS")
hospitals["region"] = hospitals["region"].str.strip()
print(hospitals["region"].unique())

#For hospital levels
print("\nHOSPITAL LEVELS")
hospitals["hospital_level"] = hospitals["hospital_level"].str.strip()
print(hospitals["hospital_level"].unique())

#For hospital names
print("\nHOSPITAL NAMES")
hospitals["hospital_name"] = hospitals["hospital_name"].str.strip()
print(hospitals["hospital_name"].unique())


#Checking for the numbers of the beds and its consistencies
print("BED COUNT CONSISTENCY:")

bed_check = (
    hospitals["number_of_beds"]
    == hospitals["beds_occupied"] + hospitals["beds_available"]
)

print(bed_check)
print("\nALL BED COUNTS CONSISTENT:")
print(bed_check.all())


#Checking for impossible bed values
print("NEGATIVE TOTAL BEDS:")
print((hospitals["number_of_beds"] < 0).sum())

print("\nNEGATIVE OCCUPIED BEDS:")
print((hospitals["beds_occupied"] < 0).sum())

print("\nNEGATIVE AVAILABLE BEDS:")
print((hospitals["beds_available"] < 0).sum())


#Checking specialities
print("SPECIALITIES:")
for i, value in enumerate(hospitals["specialities"]):
    print(f"{i}: {value}")


#Keeping the old set of specialities and its structure before standardization
hospitals["specialities_original"] = hospitals["specialities"]
hospitals["specialities_clean"] = hospitals["specialities"]


#collapsing some specialities
speciality_replacements = {
    "Dental": "Dentistry",
    "Dental Care": "Dentistry",
    "Eye Care": "Ophthalmology",
    "Emergency Care": "Emergency Medicine",
    "Accident & Emergency": "Emergency Medicine",
    "24-Hour Emergency": "Emergency Medicine",
    "Emergency": "Emergency Medicine",
    "X-Ray": "Radiology"
}

for old, new in speciality_replacements.items():
    hospitals["specialities_clean"] = (
        hospitals["specialities_clean"]
        .str.replace(old, new, regex=False)
    )

print("\nCOLLAPSED SPECIALITIES")
print(hospitals["specialities_clean"])
```

```python
#Remove Unnecessary Columns
length_of_stay = length_of_stay.drop(columns=['eid', 'vdate', 'discharged'])

#Updated Columns
print("Updated Columns")
print(length_of_stay.columns)
```

```python
#Encoding rcount, gender and facid
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

categorical_columns = ["gender", "facid"]

for col in categorical_columns:
    length_of_stay[col] = encoder.fit_transform(length_of_stay[col])


# Replace "5+" with 5
length_of_stay["rcount"] = length_of_stay["rcount"].replace("5+", 5)

# Convert to integer
length_of_stay["rcount"] = length_of_stay["rcount"].astype(int)
```
